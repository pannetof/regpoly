# CombinedF2LinearSource — XORing two MT19937s

`CombinedF2LinearSource` is the composition wrapper that XORs $J$
component F2LinearSources bitwise to widen the state space without
paying the per-step cost of a single huge generator. The combined
characteristic polynomial is the product of the components'
polynomials (when pairwise coprime), so a 2-component combination of
MT19937 with two distinct twist parameters has state size
$k_g = 2 \cdot 19937 = 39874$.

The wrapper implements `ITestable` directly (composition, not
inheritance); equidistribution kernels access its components via the
polymorphic `sources()` / `components()` accessors.

This notebook stamps two MT19937 instances with different twist
parameters $a$, combines them via `regpoly.make_combined`, and runs
the equidistribution test on the resulting 2-component combination.

See [theory/f2-linear-generators.md#combined-generators](../../theory/f2-linear-generators.md#combined-generators).


```python
# stamp:auto-generated
from regpoly import make_combined
from regpoly.core.generator import Generator
from regpoly.analyses.equidistribution_test import (
    EquidistributionTest, METHOD_HARASE,
)
INT_MAX = 2**31 - 1

# Two MT19937 components with different `a` twist constants.
gen_a = Generator.create("MTGen", L=32,
    w=32, r=624, m=397, p=31, a=0x9908B0DF)  # canonical MT19937
gen_b = Generator.create("MTGen", L=32,
    w=32, r=624, m=397, p=31, a=0xDFFFFFFF)  # alternative twist

print(f"component A: k={gen_a.k}, L={gen_a.L}")
print(f"component B: k={gen_b.k}, L={gen_b.L}")

# 2-component CombinedF2LinearSource (J = 2).
comb = make_combined(gen_a, gen_b, Lmax=32)
print(f"combined: k_g = {comb.k()}, L = {comb.L()}, J = {comb.J()}")

# Equidistribution test on the combination — MT19937 has primitive
# χ_f, so the product is primitive too and Harase applies.
test = EquidistributionTest(
    L=comb.L(),
    delta=[INT_MAX] * (comb.L() + 1),
    mse=INT_MAX,
    method=METHOD_HARASE,
)
result = test.run(comb)
print(f"SE (Σ gaps) = {result.se}")
print(f"verified    = {result.verified}")
```
